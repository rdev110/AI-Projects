### GPT-2 model configuration to answer only questions related to Python coding

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel, pipeline

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

def ask_gpt2(prompt: str):
    candidate_labels = ["python code", "not python"]
    classification = classifier(prompt, candidate_labels)
    label = classification["labels"][0]

    if label != "python code":
        return "I only answer Python code–related queries."

    inputs = tokenizer.encode(prompt, return_tensors="pt")
    outputs = model.generate(
        inputs,
        max_length=150,
        num_return_sequences=1,
        no_repeat_ngram_size=2,
        top_p=0.95,
        temperature=0.7
    )
    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
queries = [
  "Write a function to add two numbers in python",
  "How do I import pandas and read a CSV file?",
  "Tell me a story about dragons",
]

for q in queries:
  print("Question:", q)
  print("Answer:", ask_gpt2(q))
  print("-" * 60)
```
