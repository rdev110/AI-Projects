```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error


sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
!git clone "https://github.com/GeeksforgeeksDS/21-Days-21-Projects-Dataset"
```

```python
df = pd.read_csv('/content/21-Days-21-Projects-Dataset/Datasets/airline_passenger_timeseries.csv')
df.describe()
```

### 2. Exploratory Data Analysis

```python
df.plot()
plt.title('Monthly Airline Passengers (1949-1960)')
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
plt.show()
```

```python
# Convert 'Month' to datetime and set as index
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)

# Decompose the time series to visualize its components
decomposition = sm.tsa.seasonal_decompose(df['Passengers'], model='multiplicative')

fig = decomposition.plot()
fig.set_size_inches(14, 10)
plt.show()
```

### 3.Stationarity Testing

```python
def test_stationarity(timeseries):
    # Perform Dickey-Fuller test:
    print('Results of Dickey-Fuller Test:')
    dftest = adfuller(timeseries, autolag='AIC')
    dfoutput = pd.Series(dftest[0:4], index=['Test Statistic','p-value','#Lags Used','Number of Observations Used'])
    for key,value in dftest[4].items():
        dfoutput['Critical Value (%s)'%key] = value
    print(dfoutput)

test_stationarity(df['Passengers'])
```

### 4. Making the Series Stationary

```python
# 1. Apply log transformation to stabilize the variance
df_log = np.log(df['Passengers'])

# 2. Apply differencing to remove the trend
df_diff = df_log.diff().dropna()

# Plot the stationary series
df_diff.plot()
plt.title('Stationary Time Series (Log-Differenced)')
plt.show()

# Retest for stationarity
test_stationarity(df_diff)
```

### 5. Model Identification with ACF and PACF Plots

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))
plot_acf(df_diff, ax=ax1, lags=20)
plot_pacf(df_diff, ax=ax2, lags=20)
plt.show()
```

### 6. Building the ARIMA Model

```python
# Split data into training and test sets
train_data = df_log[:'1958']
test_data = df_log['1959':]

# Build ARIMA model
model = ARIMA(train_data, order=(1, 1, 1), freq='MS')
arima_result = model.fit()

# Get forecast
forecast = arima_result.get_forecast(steps=len(test_data))
forecast_ci = forecast.conf_int()

# Plot the forecast
plt.figure(figsize=(14, 7))
plt.plot(df_log, label='Original Log Data')
plt.plot(forecast.predicted_mean, label='Forecast')
plt.fill_between(forecast_ci.index, forecast_ci.iloc[:, 0], forecast_ci.iloc[:, 1], color='k', alpha=.15)
plt.title('ARIMA Model Forecast')
plt.legend()
plt.show()
```

### 7. Building a SARIMA Model for Seasonality

```python
# Build SARIMA model
# We can find the optimal P, D, Q through a grid search, but common values are 1.
sarima_model = sm.tsa.statespace.SARIMAX(train_data,
                                          order=(1, 1, 1),
                                          seasonal_order=(1, 1, 1, 12),
                                          enforce_stationarity=False,
                                          enforce_invertibility=False,
                                          freq='MS') # Explicitly set frequency to suppress warnings
sarima_result = sarima_model.fit()

# Get forecast
sarima_forecast = sarima_result.get_forecast(steps=len(test_data))
sarima_forecast_ci = sarima_forecast.conf_int()

# Plot the forecast
plt.figure(figsize=(14, 7))
plt.plot(df_log, label='Original Log Data')
plt.plot(sarima_forecast.predicted_mean, label='SARIMA Forecast', color='red')
plt.fill_between(sarima_forecast_ci.index, sarima_forecast_ci.iloc[:, 0], sarima_forecast_ci.iloc[:, 1], color='r', alpha=.15)
plt.title('SARIMA Model Forecast')
plt.legend()
plt.show()
```

### 8. Final Evaluation

```python
# Reverse the log transformation to get actual passenger numbers
original_test_data = np.exp(test_data)
sarima_predictions = np.exp(sarima_forecast.predicted_mean)

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(original_test_data, sarima_predictions))
print(f"SARIMA Model RMSE: {rmse:.2f}")

# Plot final results
plt.figure(figsize=(14, 7))
plt.plot(df['Passengers'], label='Original Data')
plt.plot(sarima_predictions, label='SARIMA Forecast', color='red')
plt.title('Final Forecast vs. Actual Data')
plt.legend()
plt.show()
```

### Q2.1 Explain the concept of stationarity and why it's important for time series modeling.

A stationary time series is one whose statistical properties do not change over time.

  - That means: The mean is constant over time.
  - The variance is constant over time (no growing fluctuations).
  - The covariance between values depends only on the lag, not on the actual time.
  - In simple words → the series “looks the same” across time, without systematic trend or changing seasonality.

Why is Stationarity Important?

**Model Assumptions**
  - Classical time series models like ARIMA, SARIMA assume stationarity.
  - Non-stationary data leads to unreliable parameter estimates and poor forecasts.

**Predictability**
  - If the process is stationary, patterns are consistent over time.
  - The model can learn from the past and apply it to the future.

**Statistical Inference**
  - Many tests (ADF, KPSS) and tools (ACF/PACF) are valid only under stationarity.

### Q2.2 Present the results of the Augmented Dickey-Fuller (ADF) test on the original data and interpret the p-value.

```python
result = adfuller(df['Passengers'])
print("ADF Statistic:", result[0])
print("p-value:", result[1])
for key, value in result[4].items():
    print(f"Critical Value ({key}): {value}")
```

### Interpretation
  - The p-value = 0.99 is much greater than 0.05.
  - We fail to reject the null hypothesis (H₀).
  - Conclusion: The original airline passenger time series is non-stationary.

### Q2.3 Apply a log transformation to the data and present the results of the ADF test after log transformation, aiming to reduce the p-value below 0.05. Discuss your findings.

```python
log_series = np.log(df['Passengers'])

# ADF test
result_log = adfuller(log_series)
print("ADF Statistic:", result_log[0])
print("p-value:", result_log[1])
for key, value in result_log[4].items():
    print(f"Critical Value ({key}): {value}")
```

### **Interpretation**
  - The p-value = 0.41 is still > 0.05, so we fail to reject H₀.
  - The log-transformed series is still non-stationary.
  - Variance is now more stable → the seasonal fluctuations look more uniform across years.

### Q2.4 Discuss the effect of differencing on the log-transformed data and present the results of the ADF test after differencing, interpreting the p-value.

```python
log_diff = log_series.diff().dropna()

# ADF test
result_log_diff = adfuller(log_diff)
print("ADF Statistic:", result_log_diff[0])
print("p-value:", result_log_diff[1])
for key, value in result_log_diff[4].items():
    print(f"Critical Value ({key}): {value}")
```

### Interpretation
**p-value = 0.071**
  - This is slightly above 0.05 → we fail to reject H₀ at the 5% level.
  - However, it is below 0.10 → meaning we can reject H₀ at the 10% level.

**Conclusion:**
  - The log-differenced series is almost stationary, but not strongly enough at the 5% confidence level.
  - It’s considered weakly stationary (stationary at 10%, but not at 5%).

### Q3. ARIMA Model Performance:
Based on the stationarity test results after log transformation (before differencing), discuss whether you would expect a non-seasonal ARIMA model to perform well on the log-transformed data. Build and evaluate a non-seasonal ARIMA model on the log-transformed data (without differencing) and compare its performance to the SARIMA model built later in the notebook.

Non-Seasonal ARIMA Model

```python
log_series = np.log(df['Passengers'])

arima_model = ARIMA(log_series, order=(2,0,2))
arima_result = arima_model.fit()

print(arima_result.summary())
forecast_arima = arima_result.forecast(steps=24)

plt.figure(figsize=(12,6))
plt.plot(log_series, label="Log Transformed Data")
plt.plot(forecast_arima, label="ARIMA Forecast", color="red")
plt.legend()
plt.show()
```

Seasonal ARIMA

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

sarima_model = SARIMAX(log_series,
                       order=(1,1,1),
                       seasonal_order=(1,1,1,12))
sarima_result = sarima_model.fit()

print(sarima_result.summary())

forecast_sarima = sarima_result.get_forecast(steps=24)
sarima_ci = forecast_sarima.conf_int()

plt.figure(figsize=(12,6))
plt.plot(log_series, label="Log Transformed Data")
plt.plot(forecast_sarima.predicted_mean, label="SARIMA Forecast", color="green")
plt.fill_between(sarima_ci.index,
                 sarima_ci.iloc[:,0],
                 sarima_ci.iloc[:,1],
                 color="lightgreen", alpha=0.4)
plt.legend()
plt.show()
```

### Compare ARIMA vs SARIMA
**Non-Seasonal ARIMA (on log data)**
  - Struggles to capture the 12-month seasonality.
  - Produces forecasts that flatten or miss peaks/troughs.
  - Model residuals show strong autocorrelation at seasonal lags → model is mis-specified.

**SARIMA**
  - Handles both trend and seasonality (via seasonal differencing and seasonal AR/MA terms).
  - Forecast aligns well with observed annual cycles.
  -Residuals look more like white noise

**Conclusion**
  - Non-seasonal ARIMA on log-transformed data (without differencing) will not perform well, because the series is still non-stationary and has strong seasonality.
  - SARIMA significantly outperforms ARIMA since it explicitly models the 12-month seasonal structure.
  - For this dataset, SARIMA (or Holt-Winters exponential smoothing) is the appropriate choice.
